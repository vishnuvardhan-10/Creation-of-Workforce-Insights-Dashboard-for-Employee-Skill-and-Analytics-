Student Attendance System




prompt : Write a code for a students attendance system  in Python using basic list, tuple, set, and OOP concepts.


```python

# Student Attendance System
# Using List, Tuple, Set and OOP

class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        self.attendance = []   # List of tuples

    def mark_attendance(self, date, status):
        # Tuple -> (date, status)
        self.attendance.append((date, status))

    def show_attendance(self):
        print(f"\nAttendance of {self.name}")
        if len(self.attendance) == 0:
            print("No attendance records.")
        else:
            for record in self.attendance:
                print(f"Date: {record[0]} | Status: {record[1]}")


class AttendanceSystem:
    def __init__(self):
        self.students = []      # List
        self.dates = set()      # Set

    def add_student(self):
        sid = input("Enter Student ID: ")
        name = input("Enter Student Name: ")

        student = Student(sid, name)
        self.students.append(student)

        print("Student Added Successfully!")

    def display_students(self):
        if len(self.students) == 0:
            print("No students available.")
        else:
            print("\nStudent List")
            print("----------------------")
            for student in self.students:
                print(f"ID: {student.student_id}  Name: {student.name}")

    def mark_attendance(self):
        sid = input("Enter Student ID: ")

        found = False

        for student in self.students:
            if student.student_id == sid:
                date = input("Enter Date (DD-MM-YYYY): ")
                status = input("Present/Absent: ")

                student.mark_attendance(date, status)

                self.dates.add(date)     # Store unique dates
                print("Attendance Marked Successfully!")
                found = True
                break

        if not found:
            print("Student not found.")

    def view_attendance(self):
        sid = input("Enter Student ID: ")

        for student in self.students:
            if student.student_id == sid:
                student.show_attendance()
                return

        print("Student not found.")

    def show_dates(self):
        print("\nAttendance Dates")
        print("-----------------")
        if len(self.dates) == 0:
            print("No dates available.")
        else:
            for date in self.dates:
                print(date)


# -----------------------------
# Main Program
# -----------------------------

system = AttendanceSystem()

while True:

    print("\n===== Student Attendance System =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Mark Attendance")
    print("4. View Attendance")
    print("5. Show Attendance Dates")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        system.add_student()

    elif choice == "2":
        system.display_students()

    elif choice == "3":
        system.mark_attendance()

    elif choice == "4":
        system.view_attendance()

    elif choice == "5":
        system.show_dates()

    elif choice == "6":
        print("Thank You!")
        break

    else:
        print("Invalid Choice! Please try again.")
```
