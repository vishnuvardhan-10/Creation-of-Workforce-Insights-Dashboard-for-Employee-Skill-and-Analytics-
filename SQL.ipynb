```python
import sqlite3

# -----------------------------------
# Create Database
# -----------------------------------
conn = sqlite3.connect("BankingDB.db")
cursor = conn.cursor()

# -----------------------------------
# Create Customer Table
# -----------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS Customer(
CustomerID INTEGER PRIMARY KEY,
Name TEXT,
AccountType TEXT,
Balance REAL,
City TEXT
)
""")

# -----------------------------------
# Create Transaction Table
# -----------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS Transactions(
TransactionID INTEGER PRIMARY KEY,
CustomerID INTEGER,
Amount REAL,
TransactionType TEXT
)
""")

# -----------------------------------
# Delete Old Data
# -----------------------------------
cursor.execute("DELETE FROM Customer")
cursor.execute("DELETE FROM Transactions")

# -----------------------------------
# Insert Customer Records
# -----------------------------------
cursor.executemany("""
INSERT INTO Customer VALUES(?,?,?,?,?)
""",[
(101,'Vishnu','Savings',25000,'Hyderabad'),
(102,'Rahul','Current',40000,'Chennai'),
(103,'Anjali','Savings',18000,'Bangalore'),
(104,'Priya','Savings',50000,'Hyderabad'),
(105,'Kiran','Current',35000,'Mumbai')
])

# -----------------------------------
# Insert Transaction Records
# -----------------------------------
cursor.executemany("""
INSERT INTO Transactions VALUES(?,?,?,?)
""",[
(1,101,5000,'Deposit'),
(2,102,3000,'Withdraw'),
(3,101,2000,'Withdraw'),
(4,104,7000,'Deposit')
])

conn.commit()

# -----------------------------------
# SELECT
# -----------------------------------
print("\n------ CUSTOMER TABLE ------")

for row in cursor.execute("SELECT * FROM Customer"):
    print(row)

# -----------------------------------
# WHERE
# -----------------------------------
print("\n------ BALANCE > 30000 ------")

for row in cursor.execute("""
SELECT * FROM Customer
WHERE Balance > 30000
"""):
    print(row)

# -----------------------------------
# UPDATE
# -----------------------------------
cursor.execute("""
UPDATE Customer
SET Balance = 30000
WHERE CustomerID=101
""")

conn.commit()

print("\n------ AFTER UPDATE ------")

for row in cursor.execute("SELECT * FROM Customer"):
    print(row)

# -----------------------------------
# DELETE
# -----------------------------------
cursor.execute("""
DELETE FROM Customer
WHERE CustomerID=103
""")

conn.commit()

print("\n------ AFTER DELETE ------")

for row in cursor.execute("SELECT * FROM Customer"):
    print(row)

# -----------------------------------
# ORDER BY
# -----------------------------------
print("\n------ ORDER BY BALANCE DESC ------")

for row in cursor.execute("""
SELECT * FROM Customer
ORDER BY Balance DESC
"""):
    print(row)

# -----------------------------------
# GROUP BY
# -----------------------------------
print("\n------ GROUP BY ACCOUNT TYPE ------")

for row in cursor.execute("""
SELECT AccountType,
COUNT(*) AS TotalCustomers,
SUM(Balance) AS TotalBalance
FROM Customer
GROUP BY AccountType
"""):
    print(row)

# -----------------------------------
# HAVING
# -----------------------------------
print("\n------ HAVING AVG(BALANCE)>30000 ------")

for row in cursor.execute("""
SELECT AccountType,
AVG(Balance)
FROM Customer
GROUP BY AccountType
HAVING AVG(Balance)>30000
"""):
    print(row)

# -----------------------------------
# INNER JOIN
# -----------------------------------
print("\n------ INNER JOIN ------")

for row in cursor.execute("""
SELECT Customer.Name,
Customer.AccountType,
Transactions.Amount,
Transactions.TransactionType
FROM Customer
INNER JOIN Transactions
ON Customer.CustomerID=Transactions.CustomerID
"""):
    print(row)

# -----------------------------------
# LEFT JOIN
# -----------------------------------
print("\n------ LEFT JOIN ------")

for row in cursor.execute("""
SELECT Customer.Name,
Customer.AccountType,
Transactions.Amount,
Transactions.TransactionType
FROM Customer
LEFT JOIN Transactions
ON Customer.CustomerID=Transactions.CustomerID
"""):
    print(row)

# -----------------------------------
# COUNT
# -----------------------------------
print("\n------ COUNT ------")

for row in cursor.execute("""
SELECT COUNT(*) FROM Customer
"""):
    print(row)

# -----------------------------------
# SUM
# -----------------------------------
print("\n------ SUM ------")

for row in cursor.execute("""
SELECT SUM(Balance) FROM Customer
"""):
    print(row)

# -----------------------------------
# AVG
# -----------------------------------
print("\n------ AVG ------")

for row in cursor.execute("""
SELECT AVG(Balance) FROM Customer
"""):
    print(row)

# -----------------------------------
# MAX
# -----------------------------------
print("\n------ MAX ------")

for row in cursor.execute("""
SELECT MAX(Balance) FROM Customer
"""):
    print(row)

# -----------------------------------
# MIN
# -----------------------------------
print("\n------ MIN ------")

for row in cursor.execute("""
SELECT MIN(Balance) FROM Customer
"""):
    print(row)

conn.close()
```
